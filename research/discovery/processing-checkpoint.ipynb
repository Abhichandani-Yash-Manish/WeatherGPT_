# WeatherGPT first processing checkpoint
Executed 12 September 2026 IST. This notebook checks local registry/build evidence and exact CSV-to-SQLite fidelity. It does not re-fetch national tables or establish scientific homogeneity, district representativeness or production readiness. The fresh route checks are separately preserved in the recheck manifest.


```python
from pathlib import Path
import csv, io, json, sqlite3, hashlib
from collections import Counter

root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "weathergpt_data/climate.py").exists())
registry = json.loads((root / "data/registry/sources.json").read_text())
by_id = {s["id"]: s for s in registry["products"]}
output = root / by_id["S25"]["processing_outputs"]["directory"]
manifest = json.loads((output / "build-manifest.json").read_text())
for name, expected in manifest["outputs"].items():
    assert hashlib.sha256((output / name).read_bytes()).hexdigest() == expected
for item in manifest["inputs"]:
    assert hashlib.sha256((root / item["path"]).read_bytes()).hexdigest() == item["sha256"]
with sqlite3.connect(output / "climate.sqlite") as con:
    print("SQLite integrity:", con.execute("PRAGMA integrity_check").fetchone()[0])
    print("Rows by source:", con.execute("SELECT source_id, COUNT(*) FROM climate_records GROUP BY source_id").fetchall())
    checked = 0
    for item in manifest["inputs"]:
        raw = list(csv.DictReader(io.StringIO((root / item["path"]).read_text())))
        rows = con.execute("SELECT source_row, source_column, native_value FROM climate_records WHERE source_id=?", (item["source_id"],)).fetchall()
        for row, col, native in rows:
            assert raw[row-2][col] == native
            checked += 1
    print("Original CSV cells matched:", checked)
    print("2024 published annual rainfall:", con.execute("SELECT value_decimal, unit, quality_flags FROM climate_records WHERE source_id='S25' AND year=2024 AND period_code='ANNUAL'").fetchone())
audit = json.loads((output / "audit.json").read_text())
print("Missing values:", audit["missing_values"])
print("Rainfall totals exceeding rounding bound:", len(audit["rainfall_totals_beyond_rounding_bound"]))
print("Registry entries:", len(registry["products"]))
print("Processing states:", dict(Counter(s["processing_state"] for s in registry["products"])))
```
